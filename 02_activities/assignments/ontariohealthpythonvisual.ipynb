```python
import pandas as pd
import matplotlib.pyplot as plt
import openpyxl

file_path = "/Users/andradacretu/Documents/Documents - Andrada’s MacBook Pro/visualization/02_activities/assignments/ontario-health-premium-rates-en.xlsx"
excel_data = pd.ExcelFile(file_path)

df = excel_data.parse('ontario-health-premium-rates-en')

df.columns = ['Income', 'Premium']


df = df[df['Income'] != 'Up to 20,000']


df['Income'] = pd.to_numeric(df['Income'], errors='coerce')
df['Premium'] = pd.to_numeric(df['Premium'], errors='coerce')


df = df.dropna()

# Create the line plot
plt.figure(figsize=(10, 6))
plt.plot(df['Income'], df['Premium'], marker='o', linestyle='-', linewidth=2)
plt.title('Ontario Health Premium vs. Taxable Income', fontsize=14)
plt.xlabel('Individual Taxable Income (CAD)', fontsize=12)
plt.ylabel('Ontario Health Premium (CAD)', fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
```
